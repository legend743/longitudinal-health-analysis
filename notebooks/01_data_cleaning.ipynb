```python
patients["gender"]=patients["gender"].replace({
    "male":"Male",
    "M":"Male",
    "m":"Male",
    "F":"Female"
})
patients.loc[patients["age"]>100,"age"]=np.nan
print(patients["gender"].value_counts())
print(patients.isnull().sum())
```

```python
patients["age"] = patients["age"].fillna(patients["age"].median())

patients["gender"] = patients["gender"].fillna(patients["gender"].mode()[0])

print(patients.isnull().sum())
```
